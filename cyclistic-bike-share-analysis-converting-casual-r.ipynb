##  📌 Business Problem

Cyclistic is a bike-share company aiming to increase profitability by converting casual riders into annual members.

The key business question:
**How do annual members and casual riders use Cyclistic bikes differently?**

Understanding these differences will help design targeted marketing strategies to convert casual users into long-term subscribers.

Problem

Cyclistic is a bike-share company aiming to increase profitability by converting casual riders into annual members.

The key business question:
**How do annual members and casual riders use Cyclistic bikes differently?**

Understanding these differences will help design targeted marketing strategies to convert casual users into long-term subscribers.

## 📊 Data Description

- Source: Divvy (Cyclistic) public dataset
- Period Used: 2 months (due to storage constraints)
- Data Type: Trip-level ride data

### Key Columns:
- ride_id
- rideable_type
- started_at / ended_at
- start_station / end_station
- member_casual

### Limitations:
- Only 2 months used (not full-year seasonality)
- No demographic or personal user data

## 🧹 Data Cleaning & Preparation

Steps performed:
- Merged datasets
- Removed duplicates and null values
- Converted date columns
- Created new features:
  - ride_length
  - day_of_week
  - hour

### 🚴 Ride Length Distribution

Casual riders tend to take longer rides compared to members.

### ⏱ Average Ride Duration

Casual riders spend significantly more time per ride than members.

### 📅 Weekly Usage Patterns

- Members ride more during weekdays → commuting behavior  
- Casual riders peak on weekends → leisure usage

### 🕒 Time-of-Day Patterns

- Members: Peak during morning & evening commute hours  
- Casual riders: Midday usage (flexible/leisure rides)

### 🔥 Ride Density Heatmap

Clear behavioral segmentation:
- Members → structured usage
- Casual → flexible patterns

## 🔍 Key Insights

1. Casual riders take longer rides but less frequently  
2. Members take shorter rides but more consistently  
3. Members dominate weekday usage (commuting)  
4. Casual riders dominate weekend usage (leisure)  
5. Members show clear peak hours; casual riders are more flexible  

👉 Conclusion:
Casual riders use bikes for experience, while members use them for routine transportation.

## 💡 Business Recommendations

### 1. Convert Weekend Riders
Target casual riders with weekend-heavy usage using weekday ride incentives.

### 2. Introduce Membership Trials
Offer discounted or trial memberships to frequent casual users.

### 3. Digital Marketing Campaigns
Use email/social ads emphasizing:
- Cost savings
- Convenience for daily commuting

## ⚠️ Limitations & Next Steps

### Limitations:
- Only 2 months of data used
- No demographic insights

### Next Steps:
- Expand to 12 months for seasonality analysis
- Integrate weather data for deeper insights
- Use BigQuery for scalable analysis

## ✅ Conclusion

There is a clear behavioral difference between casual riders and members.

By converting casual riders into habitual users, Cyclistic can significantly increase revenue and long-term growth.
